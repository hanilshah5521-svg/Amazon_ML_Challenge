```python
import re
import unicodedata
import pandas as pd


def normalize_text(text):
    """
    Basic text normalization:
    - handle missing values
    - lowercase
    - remove accents
    - remove punctuation
    - normalize whitespace
    """

    if pd.isna(text):
        return ""

    # Convert to string
    text = str(text)

    # Lowercase
    text = text.lower()

    # Remove accents / diacritics
    text = unicodedata.normalize("NFKD", text)
    text = "".join(
        char for char in text
        if not unicodedata.combining(char)
    )

    # Replace punctuation/special characters with spaces
    text = re.sub(r"[^a-z0-9\u0900-\u097f\u0b80-\u0bff]+", " ", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text


def normalize_name(text):
    """
    Normalize business names.
    """
    return normalize_text(text)


def normalize_address(text):
    """
    Normalize business addresses.
    """
    return normalize_text(text)
```

```python
import re
import unicodedata
import pandas as pd


# Common address abbreviations
ADDRESS_ABBREVIATIONS = {
    "street": "st",
    "road": "rd",
    "avenue": "ave",
    "boulevard": "blvd",
    "lane": "ln",
    "drive": "dr",
    "highway": "hwy",
    "parkway": "pkwy",
    "place": "pl",
    "court": "ct",
    "circle": "cir",
    "terrace": "ter",
    "square": "sq",
}


def normalize_text(text):
    if pd.isna(text):
        return ""

    text = str(text).lower()

    # Unicode normalization
    text = unicodedata.normalize("NFKD", text)

    # Remove combining accents
    text = "".join(
        char for char in text
        if not unicodedata.combining(char)
    )

    # Replace punctuation with spaces
    text = re.sub(
        r"[^a-z0-9\u0900-\u097f\u0b80-\u0bff]+",
        " ",
        text
    )

    # Normalize spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text


def normalize_name(text):
    return normalize_text(text)


def normalize_address(text):
    text = normalize_text(text)

    if not text:
        return ""

    # Normalize common address words
    words = text.split()

    words = [
        ADDRESS_ABBREVIATIONS.get(word, word)
        for word in words
    ]

    return " ".join(words)
```

```python
def jaccard_similarity(text1, text2):
    if not text1 or not text2:
        return 0.0

    set1 = set(text1.split())
    set2 = set(text2.split())

    if not set1 or not set2:
        return 0.0

    return len(set1 & set2) / len(set1 | set2)
```

```python
results = []

for _, row in sample_gt.iterrows():

    s1 = s1_dict[row["source1_entity_id"]]

    matched_ids = str(row["matched_entity_ids"])

    if matched_ids == "nan" or matched_ids == "":
        continue

    for match_id in matched_ids.split(","):

        match_id = match_id.strip()

        if match_id.startswith("S2-"):
            match = s2_dict[match_id]
        else:
            match = s3_dict[match_id]

        name1 = normalize_name(s1["business_name"])
        name2 = normalize_name(match["business_name"])

        addr1 = normalize_address(s1["business_address"])
        addr2 = normalize_address(match["business_address"])

        results.append({
            "s1_id": row["source1_entity_id"],
            "match_id": match_id,
            "name_jaccard": jaccard_similarity(name1, name2),
            "address_jaccard": jaccard_similarity(addr1, addr2)
        })

jaccard_results = pd.DataFrame(results)

print("Total true match pairs:", len(jaccard_results))
```

```python
thresholds = [0.3, 0.4, 0.5, 0.6, 0.7, 0.8, 0.9]

print("NAME JACCARD")
print("=" * 40)

for t in thresholds:
    recall = (jaccard_results["name_jaccard"] >= t).mean()
    print(f"Threshold {t:.1f}: {recall:.2%}")


print("\nADDRESS JACCARD")
print("=" * 40)

for t in thresholds:
    recall = (jaccard_results["address_jaccard"] >= t).mean()
    print(f"Threshold {t:.1f}: {recall:.2%}")


print("\nNAME OR ADDRESS")
print("=" * 40)

for t in thresholds:
    recall = (
        (jaccard_results["name_jaccard"] >= t) |
        (jaccard_results["address_jaccard"] >= t)
    ).mean()

    print(f"Threshold {t:.1f}: {recall:.2%}")
```

```python
def char_ngrams(text, n=3):
    if not text:
        return set()
    
    text = text.replace(" ", "")
    
    if len(text) < n:
        return {text}
    
    return {
        text[i:i+n]
        for i in range(len(text) - n + 1)
    }


def char_jaccard(text1, text2, n=3):
    grams1 = char_ngrams(text1, n)
    grams2 = char_ngrams(text2, n)
    
    if not grams1 or not grams2:
        return 0.0
    
    return len(grams1 & grams2) / len(grams1 | grams2)
```

```python
# ============================================================
# CHARACTER TF-IDF — ADDRESS EXPERIMENT
# ============================================================

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd


# ------------------------------------------------------------
# STEP 1: Prepare address text
# ------------------------------------------------------------

candidate_address_texts = (
    candidate_pool["business_address"]
    .fillna("")
    .apply(normalize_address)
    .tolist()
)

s1_address_texts = (
    s1_small["business_address"]
    .fillna("")
    .apply(normalize_address)
    .tolist()
)


# ------------------------------------------------------------
# STEP 2: Create character-level TF-IDF
# ------------------------------------------------------------

address_vectorizer = TfidfVectorizer(
    analyzer="char_wb",
    ngram_range=(3, 5),
    min_df=1
)

candidate_address_matrix = address_vectorizer.fit_transform(
    candidate_address_texts
)

s1_address_matrix = address_vectorizer.transform(
    s1_address_texts
)

print("Candidate address matrix:",
      candidate_address_matrix.shape)

print("S1 address matrix:",
      s1_address_matrix.shape)


# ------------------------------------------------------------
# STEP 3: Calculate cosine similarity
# ------------------------------------------------------------

address_tfidf_similarity = cosine_similarity(
    s1_address_matrix,
    candidate_address_matrix
)

print(
    "Address similarity matrix:",
    address_tfidf_similarity.shape
)


# ------------------------------------------------------------
# STEP 4: Get TF-IDF scores for TRUE MATCHES
# ------------------------------------------------------------

address_tfidf_true_scores = []

for s1_id, candidate_id in true_pairs:

    i = s1_index[s1_id]
    j = candidate_index[candidate_id]

    score = address_tfidf_similarity[i, j]

    address_tfidf_true_scores.append(score)


address_tfidf_true_scores = pd.Series(
    address_tfidf_true_scores
)


# ------------------------------------------------------------
# STEP 5: Basic statistics
# ------------------------------------------------------------

print("\nTrue pairs:",
      len(address_tfidf_true_scores))

print("\nScore statistics:")
print(address_tfidf_true_scores.describe())


# ------------------------------------------------------------
# STEP 6: Test different thresholds
# ------------------------------------------------------------

thresholds = [
    0.3,
    0.4,
    0.5,
    0.6,
    0.7,
    0.8,
    0.9
]

print("\nCHARACTER TF-IDF — ADDRESS")
print("=" * 45)

for t in thresholds:

    recall = (
        address_tfidf_true_scores >= t
    ).mean()

    print(
        f"Threshold {t:.1f}: "
        f"{recall:.2%}"
    )
```

```python
# ============================================================
# COMBINED BLOCKER — FULL EVALUATION
# ============================================================

# Thresholds we want to test
thresholds = [0.3, 0.4, 0.5, 0.6, 0.7]

results = []

for t in thresholds:

    # --------------------------------------------------------
    # Select candidates using ANY of our four signals
    # --------------------------------------------------------
    
    selected = combined_results[
        (combined_results["name_word"] >= t) |
        (combined_results["addr_word"] >= t) |
        (combined_results["name_char"] >= t) |
        (combined_results["addr_char"] >= t)
    ].copy()

    
    # --------------------------------------------------------
    # Convert selected candidates into (S1, candidate) pairs
    # --------------------------------------------------------
    
    selected_pairs = set(
        zip(
            selected["s1_id"],
            selected["candidate_id"]
        )
    )

    
    # --------------------------------------------------------
    # TRUE candidates recovered
    # --------------------------------------------------------
    
    true_recovered = len(
        true_pairs & selected_pairs
    )

    
    # --------------------------------------------------------
    # FALSE candidates
    # --------------------------------------------------------
    
    false_candidates = (
        len(selected_pairs)
        - true_recovered
    )

    
    # --------------------------------------------------------
    # Candidate recall
    # --------------------------------------------------------
    
    recall = (
        true_recovered / len(true_pairs)
        if len(true_pairs) > 0
        else 0
    )

    
    # --------------------------------------------------------
    # Candidate precision
    # --------------------------------------------------------
    
    precision = (
        true_recovered / len(selected_pairs)
        if len(selected_pairs) > 0
        else 0
    )

    
    # --------------------------------------------------------
    # Average candidates per S1
    # --------------------------------------------------------
    
    avg_candidates = (
        len(selected_pairs) / len(s1_small)
    )

    
    results.append({
        "threshold": t,
        "candidates": len(selected_pairs),
        "true_recovered": true_recovered,
        "false_candidates": false_candidates,
        "recall": recall,
        "precision": precision,
        "avg_candidates_per_s1": avg_candidates
    })


# ------------------------------------------------------------
# Create result table
# ------------------------------------------------------------

blocker_evaluation = pd.DataFrame(results)


# ------------------------------------------------------------
# Display nicely
# ------------------------------------------------------------

print(
    blocker_evaluation.to_string(
        index=False,
        formatters={
            "recall": "{:.2%}".format,
            "precision": "{:.2%}".format,
            "avg_candidates_per_s1": "{:.2f}".format
        }
    )
)
```
